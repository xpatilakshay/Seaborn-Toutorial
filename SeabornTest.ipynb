## Part A: Relational Plots

### Q1. Scatterplot (advanced usage)
Load the penguins dataset. Create a scatterplot of flipper_length_mm vs body_mass_g:

Differentiate species with hue.

Differentiate sex with style.

Use bill_length_mm as size.

Set alpha=0.6.

Change palette to "Set2".

Add a legend in top-left, shrink it, and give a title "Penguins Distribution".

```python
import seaborn as sns
import matplotlib.pyplot as plt

pen = sns.load_dataset("penguins")
sns.scatterplot(data=pen,x="flipper_length_mm",y="body_mass_g",hue="species",style="sex",size="bill_length_mm",alpha=0.6,palette="Set2")
plt.legend(loc="upper left",fontsize=7,title_fontsize=9,title="Legend",markerscale=0.7)
# plt.legend(prop={"size":7})
plt.title("Penguins Distribution")
plt.show()
```

### Q2. Lineplot
Load the fmri dataset.

Plot timepoint vs signal.

Use hue="region", style="event".

Add ci="sd" (show standard deviation).

Change line style (dashes=False).

Use a custom color palette.

Add markers (marker="o").

```python
fmri = sns.load_dataset("fmri")
color = sns.color_palette("Set2",6)
sns.lineplot(data=fmri,x="timepoint",y="signal",hue="region",style="event",ci="sd",dashes=False,palette=color,marker="o",markers=True)
plt.show()
```

## Part B: Distribution Plots

### Q3. Histplot + Rugplot
Load tips dataset. Plot histogram of total_bill:

Add kde=True.

Overlay a rugplot.

Set bins=20, element="step".

Change color palette to "rocket".

Normalize the histogram (stat="density").

```python
tips = sns.load_dataset("tips")
sns.histplot(data=tips,x="total_bill",kde=True,bins=20,element="step",color="purple",stat="density")
sns.rugplot(data=tips,x="total_bill",color="black")
plt.show()
```

### Q4. KDEplot (extra)
Plot KDE of tip grouped by sex:

Shade the area.

Use different line styles for each group.

Add fill transparency (alpha=0.5).


```python
sns.kdeplot(data=tips,x="tip",hue="sex",alpha=0.5,fill=True)
plt.show()
```

## Part C: Categorical Plots

### Q5. Barplot
From tips:

Barplot of average tip per day.

Use hue="sex".

Change estimator to np.median.

Add error bars with errorbar=("ci", 95).

Change palette to "muted".

```python
import numpy as np
sns.barplot(data=tips,x="day",y="tip",hue="sex",estimator=np.median,errorbar=("ci",95),palette="muted")
plt.show()
```

### Q6. Countplot

Count of smokers split by sex (hue="sex").

Order bars by frequency.

Change color palette to "coolwarm".

```python
frequency = tips["smoker"].value_counts().index
sns.countplot(data=tips,x="smoker",hue="sex",order=frequency,palette="coolwarm")
plt.show()
```

### Q7. Boxplot

Boxplot of total_bill across day.

Split by smoker status (hue="smoker").

Customize outliers (flierprops={"marker":"d","color":"red","markersize":6}).

Set orientation to horizontal.

```python
sns.boxplot(data=tips,x="day",y="total_bill",flierprops={"marker":"d","color":"red","markersize":6},orient="v")
plt.show()
```

### Q8. Violinplot

Violinplot of tip across sex.

Split by smoker.

Inner style = "quartile".

Use bw=0.3 (bandwidth).


```python
sns.violinplot(data=tips,x="tip",y="smoker",hue="sex",split=True,inner="quartile",bw=0.3)
plt.show()
```

### Q9. Stripplot + Swarmplot

Overlay a stripplot on a boxplot (jitter=True, alpha=0.5).

Create a swarmplot of tips per day with hue="sex".

Change size of points.

```python
sns.boxplot(data=tips,x="day",y="total_bill",flierprops={"marker":"d","color":"red","markersize":6},orient="v")
sns.stripplot(data=tips,x="day",y="total_bill",jitter=True,alpha=0.5)
plt.show()
sns.swarmplot(data=tips,x="day",y="tip",hue="sex",size=7)
plt.show()
```

## Part D: Matrix Plots

### Q10. Heatmap
Using flights dataset:

Pivot (month vs year).

Create a heatmap with:

annot=True, fmt="d"

linewidths=0.5, linecolor="black"

cmap="YlGnBu"

cbar=True, with cbar_kws={"shrink":0.7,"orientation":"horizontal"}

Fix value limits: vmin=100, vmax=700

square=True

```python
flights = sns.load_dataset("flights").pivot(index="month",columns="year",values="passengers")
corr = flights.corr(numeric_only=True)
sns.heatmap(corr,annot=True,fmt=".2f",linewidth=0.5,linecolor="black",cmap="YlGnBu",
            cbar=True,cbar_kws={"shrink":0.7,"orientation":"horizontal"},
            vmin=-1,vmax=1)
plt.show()
```

### Q11. Clustermap

Use the same pivot table.

Add annot=True.

Use cmap="mako".

Apply standard_scale=1.

Rotate tick labels for better readability.

```python
g=sns.clustermap(flights,annot=True,cmap="mako",standard_scale=1)
g.ax_heatmap.set_xticklabels(g.ax_heatmap.get_xticklabels(), rotation=45)
plt.show()
```

## Part E: Multivariate Plots

### Q12. Pairplot
Use iris dataset.

Plot pairplot with hue="species".

Diagonal: show KDE instead of hist.

Kind = scatter.

Use palette "husl".

Add markers (markers=["o","s","D"]).

```python
iris = sns.load_dataset("iris")
sns.pairplot(data=iris,hue="species",diag_kind="kde",kind="scatter",palette="husl",markers=["o","s","d"])
plt.show()
```

## Part F: Theme & Aesthetics

### Q13. Custom Styling

Set global theme: sns.set_style("whitegrid").

Remove left and top spines (sns.despine).

Use sns.set_context("talk",font_scale=1.2) for larger plots.

Apply custom color palette (manual hex colors like ["#FF6F61","#6B5B95","#88B04B"]).

## Part G: Final Challenge – Combined Dashboard

### Q14. Subplot Dashboard
Create a 3x1 subplot layout:

Countplot of day in tips dataset.

Violinplot of tip vs sex split by smoker.

Heatmap of flights pivot table.

Requirements:

Use plt.subplots(figsize=(14,10)).

Apply consistent sns.set_context("notebook").

Apply one shared palette.

Add subplot titles.

Adjust spacing (plt.tight_layout()).

```python
fig, axes = plt.subplots(3,1,figsize=(14,14))

sns.set_style("whitegrid")
sns.set_context("notebook", font_scale=1.2)
palette = ["#FF6F61", "#6B5B95", "#88B04B"]
sns.set_palette(palette)

sns.countplot(data=tips,x="day",ax=axes[0])
axes[0].set_title("Count of Records per Day")
sns.despine(ax=axes[0],left=True,top=True)

sns.violinplot(data=tips,x="sex",y="tip",hue="smoker",split=True,ax=axes[1])
axes[1].set_title("Tip Distribution by Sex and Smoker")
sns.despine(ax=axes[1],left=True,top=True)

sns.heatmap(corr,annot=True,fmt=".2f",ax=axes[2])
axes[2].set_title("Monthly Passengers Over Years")
sns.despine(ax=axes[2],left=True,top=True)

plt.tight_layout()
plt.show()
```
