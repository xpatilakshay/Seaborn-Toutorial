## 1. sns.histplot

👉 What it does:

Creates a histogram (counts of data values in bins).

Can overlay a KDE curve (density estimation).

Useful for understanding how data is distributed (normal, skewed, outliers, multimodal).

## Important Parameters:

x or y: Variable to plot.

bins: Number of bins (controls granularity).

hue: Separate histograms by category.

multiple: Controls how multiple categories are displayed:

"stack" (stacked bars)

"dodge" (side-by-side bars)

"fill" (proportionally stacked, total height=1)

"layer" (overlap on top of each other)

element: Shape of bars ("bars", "step", "poly").

kde=True: Add a KDE line on top.

```python
sns.histplot(x="total_bill",data=tips,bins=20,kde=True,element="step")
plt.title("Tota Bills")
plt.show()
```

## 📌 What is kdeplot?

A smoothed curve that estimates the probability density of a continuous variable.

Unlike histograms (which depend on bin size), KDE is smooth and continuous.

Helps visualize distribution patterns more clearly.

```python
sns.kdeplot(data=tips,x="total_bill",fill=True,color="skyblue",hue="sex")
plt.show()
```

```python
sns.kdeplot(data=tips,x="total_bill",y="tip",fill=True,color="skyblue",hue="sex",cmap="mako")
plt.show()
```

## Empirical Cumulative Distribution Function (ECDF) at a given value x tells us the proportion (or percentage) of data points that are less than or equal to x.

### ✅ Example

Suppose we have the dataset: [2, 3, 3, 5, 10].

At x = 3,

Data points ≤ 3 are [2, 3, 3] → 3 out of 5 values

ECDF(3) = 3/5 = 0.6 (60%)

Interpretation: 60% of the data is less than or equal to 3.

At x = 5,

Data points ≤ 5 are [2, 3, 3, 5] → 4 out of 5 values

ECDF(5) = 4/5 = 0.8 (80%)

Interpretation: 80% of the data is less than or equal to 5.

### By default (complementary=False), ECDF shows the proportion of observations less than or equal to a given value.

Example: If x = 200 and ECDF value is 0.6, then 60% of the data points are ≤ 200.

When complementary=True, the ECDF shows the proportion of observations greater than or equal to a given value.

Example: If x = 200 and complementary ECDF value is 0.4, then 40% of the data points are ≥ 200.

```python
# Basic ECDF with tips dataset

sns.ecdfplot(data=tips, x="total_bill")
plt.show()
```

```python
# ECDF with groups (compare smokers vs. non-smokers)

sns.ecdfplot(data=tips, x="total_bill",hue="smoker")
plt.show()
```

# 🔹 sns.regplot()

### Use case:

Best for quick, simple regression plots.

Adds a regression line + scatterplot in one axis only.

Good when you’re exploring two variables only.

### Key Features:

Works with both continuous & discrete x-values.

Can add jitter, lowess smoothing, disable confidence interval.

Integrated inside a single plt axis → easy for combining with other plots.

```python
sns.regplot(data=tips,x="total_bill",y="tip",ci=None)
plt.show()
```

# 🔹 sns.lmplot()

### Use case:

Best for multi-variable regression analysis.

Can create facet grids (row/column subplots).

Automatically handles grouping with hue, row, col.

Great when you want separate regression lines per category.

### Key Features:

High-level wrapper built on top of FacetGrid.

Can compare multiple groups in one line of code.

More flexible for statistical exploration of data.

```python
sns.lmplot(data=tips,x="total_bill",y="tip",hue="sex",col="smoker",ci=60)
plt.show()
```
